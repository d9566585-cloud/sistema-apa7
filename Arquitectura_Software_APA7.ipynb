# ¿Qué es la Arquitectura de Software?

La arquitectura de software es la estructura que define cómo se organizan y comunican los componentes de un sistema.

Una buena arquitectura permite:

- Escalabilidad
- Mantenimiento sencillo
- Mejor rendimiento
- Organización del sistema
- Facilidad para futuras mejoras

La arquitectura es fundamental en sistemas modernos porque ayuda a controlar la complejidad del software.

# Tipos de Arquitectura de Software

## Arquitectura Monolítica

Todo el sistema funciona como una sola aplicación integrada.

### Ventajas
- Fácil desarrollo inicial
- Menor complejidad

### Desventajas
- Difícil escalabilidad
- Mantenimiento complicado

---

## Arquitectura de Microservicios

El sistema se divide en pequeños servicios independientes.

### Ventajas
- Escalable
- Flexible
- Mejor mantenimiento

### Desventajas
- Mayor complejidad
- Comunicación entre servicios

---

## Arquitectura Serverless

La infraestructura es administrada automáticamente por servicios cloud.

### Ventajas
- Escalado automático
- Menor administración

### Desventajas
- Dependencia del proveedor

# Descripción del Sistema APA 7

El sistema desarrollado tiene como objetivo analizar documentos académicos y corregir automáticamente errores relacionados con las normas APA 7.

El sistema permite:

- Detectar errores de formato
- Verificar márgenes y tipografía
- Analizar referencias bibliográficas
- Generar correcciones automáticas
- Normalizar documentos académicos

Este proyecto busca reducir errores manuales y optimizar el tiempo de revisión de trabajos académicos.

````python
import time

print("==========================================")
print(" SISTEMA APA 7 - SIMULACIÓN ARQUITECTURA ")
print("==========================================\n")

time.sleep(1)
print("👤 Usuario accede al sistema")

time.sleep(1)
print("📄 Usuario carga documento .docx")

time.sleep(1)
print("🔍 Analizador de documento iniciado")

time.sleep(1)
print("⚠️ Errores APA detectados:")
print("   - Interlineado incorrecto")
print("   - Márgenes incorrectos")
print("   - Referencias mal estructuradas")

time.sleep(2)
print("\n🛠️ Aplicando correcciones automáticas...")

time.sleep(1)
print("✅ Fuente corregida")
print("✅ Márgenes normalizados")
print("✅ Títulos organizados")
print("✅ Referencias generadas")

time.sleep(2)
print("\n📥 Generando documento corregido...")

time.sleep(1)
print("🎉 Documento APA 7 normalizado correctamente")
````
